```python
%sql
Create Catalog if not exists Career_Copilot_dev
```

```python
%sql
Create schema if not exists Career_Copilot_dev.common
```

```python
%sql
CREATE SCHEMA IF NOT EXISTS Career_Copilot_dev.main
```

```python
%sql
CREATE TABLE IF NOT EXISTS Career_Copilot_dev.common.users (
  user_id STRING NOT NULL,
  
  -- User profile information
  email STRING,
  first_name STRING,
  last_name STRING,
  
  -- Resume management
  active_resume_id STRING,              -- ID of the currently active resume
  resume_folder_path STRING NOT NULL,   -- Dedicated folder path for user's resumes
                                         -- Example: /Volumes/career_copilot_dev/common/resumes/user_<user_id>/
  
  -- Account metadata
  account_status STRING,                 -- active, inactive, suspended
  created_at TIMESTAMP,
  updated_at TIMESTAMP,
  
  -- Primary key
  CONSTRAINT users_pk PRIMARY KEY (user_id)
)
USING DELTA
COMMENT 'User registry with profile info and resume folder paths';
```

```python
%sql
-- Grant USE CATALOG permission to all account users (includes app service principals)
GRANT USE CATALOG ON CATALOG Career_Copilot_dev TO `account users`;

-- Grant MODIFY permission to allow creating/modifying objects
GRANT MODIFY ON CATALOG Career_Copilot_dev TO `account users`;
```

```python
%sql
-- Grant USE SCHEMA permissions
GRANT USE SCHEMA ON SCHEMA Career_Copilot_dev.common TO `account users`;
GRANT USE SCHEMA ON SCHEMA Career_Copilot_dev.main TO `account users`;

-- Grant MODIFY to allow creating/modifying tables in schemas
GRANT MODIFY ON SCHEMA Career_Copilot_dev.common TO `account users`;
GRANT MODIFY ON SCHEMA Career_Copilot_dev.main TO `account users`;
```

```python
%sql
-- Grant SELECT and MODIFY on the users table
GRANT SELECT ON TABLE Career_Copilot_dev.common.users TO `account users`;
GRANT MODIFY ON TABLE Career_Copilot_dev.common.users TO `account users`;

-- Grant SELECT and MODIFY on all future tables in these schemas
GRANT SELECT, MODIFY ON SCHEMA Career_Copilot_dev.common TO `account users`;
GRANT SELECT, MODIFY ON SCHEMA Career_Copilot_dev.main TO `account users`;
```
